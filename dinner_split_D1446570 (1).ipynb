## 三、問題回答

### 問題一：為什麼 `"4"` 要先轉成數字，才能用於本題的分攤計算？

在 Python 中，加上引號的 `"4"` 是字串（`str`）型別，電腦會把它當成純文字看待，而不是具有數學意義的數值。字串無法直接進行除法（`/`）等數學運算；如果直接拿應付總額去除以 `"4"`，程式會發生 `TypeError` 型別錯誤而無法執行。因此，必須先使用 `int("4")` 將其轉換為整數（`int`）`4`，才能正確計算每人分攤的金額。

### 問題二：如果先扣折價券，再依剩餘餐費計算服務費，第一組的應付總額會變成多少？為什麼和題目規定的結果不同？

- 改變順序後的計算過程：
i. 先扣折價券：1200 - 100 = 1100 元（折價後餐費）
ii. 再算 10% 服務費：1100 × 0.10 = 110 元
iii. 應付總額：1100 + 110 = 1210 元
- 差異原因：題目規定服務費必須以「原始餐費（1,200 元）」計算，因此服務費是 1200 × 0.10 = 120 元，加上餐費並扣除折價券後，總額為 1,220 元。若先扣除折價券，計算服務費的基準金額（母數）就從 1,200 元縮小為 1,100 元，導致服務費少了 10 元（變成 110 元），最終應付總額也就少了 10 元。這在實務上也符合多數餐廳「服務費依原價計算」的收費慣例。
